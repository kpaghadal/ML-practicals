**Exercise 1: Implement Simple Linear Regression**

Use the computers.csv dataset and perform the following activities:

Dataset contains the details of time taken to repair a computer, given the number of units to be repaired.

```python
computers = pd.read_csv("data/computers.csv")
```

```python
computers.info()
```

```python
computers.columns
```

```python
mean_value=computers['Minutes'].mean()
mean_value
```

```python
plt.scatter(computers['Units'],computers['Minutes'])
plt.xlabel('Units')
plt.ylabel('Minutes')
```

Create Models

```python
# Model 0
Model0 = computers['Minutes'].mean()
# Model 1
Model1 = 10+12*computers['Units']
# Model 2
Model2 = 6+18*computers['Units']
# Model 3
Model3 = 4.16+15.50*computers['Units']
```

```python
computers['Model0'] = Model0
computers['Model1'] = Model1
computers['Model2'] = Model2
computers['Model3'] = Model3
```

```python
computers.columns
```

```python
computers
```

```python
fig, ax = plt.subplots()
ax.scatter(x="Units",y="Minutes",data=computers,label='actual repair time')
ax.plot(computers['Units'],computers['Model0'],color="red",label="Model0")
ax.plot(computers['Units'],computers['Model1'],color="green",label="Model1")
ax.plot(computers['Units'],computers['Model2'],color="black",label="Model2")
ax.plot(computers['Units'],computers['Model3'],color="purple",label="Model3")
ax.set_ylabel("Minutes")
ax.set_xlabel("Units")
ax.set_title("Speculated Models")
ax.legend()
```

Bulid scikit learn model - Linear Regression

```python
# 1. Build Regression model using Scikit-Learn Library(Model 3).
#   a. Set the 'Units' column as the input data or predictor column
#   b. Set the 'Minutes' column as the output data
#   c. Importing the required class -  LinearRegression
#   d.Creating a linear regression model
#   e.Fitting the model to the data

X = computers[['Units']]
y = computers['Minutes']
from sklearn.linear_model import LinearRegression
# Object
model = LinearRegression()
model.fit(X,y)
```

```python
#   f. Fetching intercept and coefficient
print("Intercept : ",model.intercept_)
print("Coefficient : ",model.coef_)
```

```python
# Accuracy
Rsq = model.score(computers[["Units"]],y)
Rsq
```

**Exercise 2: Multiple Linear Regression Model**

std_marks_data.csv

To understand Multiple Linear Regression, let us consider the student marks dataset.
This model helps us to predict the marks of the students using the previous dataset.
The inputs are hours(time spent to study) ,age(present age) and internet(weather the internet is available or not), by using these 3 input fields we are predicting the marks of the students.

```python
Student = pd.read_csv("data/std_marks_data.csv")
```

```python
Student.info()
```

```python
Student.columns
```

```python
Student
```

```python
Student.isna().sum()
```

```python
Student['hours'] = Student['hours'].fillna(Student['hours'].mean())
# Student.isna().sum()
```

```python
X=Student.iloc[:,:-1]
```

```python
X
```

```python
y=Student.iloc[:,-1]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3)
```

```python
X_train.shape
# y_train.shape
# X_test.shape
# y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
print("Intercept : ",model.intercept_)
print("Coefficient : ",model.coef_)
# print("Intercept : ",model.intercept_,"\nCoefficient : ",model.coef_)
```

```python
hours=int(input("Enter study hours : "))
age=int(input("Enter age : "))
internet=int(input("Enter internet is available or not (0 or 1) : "))
data=[[hours,age,internet]]
print(model.predict(data))
```

**Exercise 3: Build Logistic Regression Model**

Consider a coronary heart disease dataset (chd_data.csv) which lists the age in years ('age') and the presence/absence of evidence of significant coronary heart disease ('chd') for 100 patients.

The variable chd = 0 indicates the absence of coronary heart disease, whereas chd=1 indicates the presence of coronary heart disease.

```python
chd = pd.read_csv("chd_data.csv")
```

```python
chd.info()
```

```python
chd
```

```python
chd.isna().sum()
```

```python
plt.scatter('age','chd',data=chd)
```

```python
X = chd[['age']]
y = chd['chd']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=0)
```

```python
X_train.shape
# y_train.shape
# X_test.shape
# y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train,y_train)
```

```python
print("Intercept : ",model.intercept_)
print("Coefficient : ",model.coef_)
```

```python
age=int(input("Enter age : "))
data=[[age]]
print("\nPredicted target i.e. 'chd' value : ",model.predict(data))
```

```python
print("Accuracy score of the model on training data : ",model.score(X_train,y_train))
print("Accuracy score of the model on test data : ",model.score(X_test,y_test))
```
